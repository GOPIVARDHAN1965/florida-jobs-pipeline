```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time

# Keywords to search for
keywords = [
    "data", "analyst", "developer", "engineer", "python", "sql",
    "bi", "cloud", "aws", "etl", "software", "finance", "financial"
]

# Setup Chrome driver
options = webdriver.ChromeOptions()
options.add_argument("--start-maximized")
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

# Open job listings
driver.get("https://jobs.myflorida.com/search/?searchby=location")

# Wait for page to load
time.sleep(3)

page_num = 1
while True:
    print(f"\n--- PAGE {page_num} ---\n")
    time.sleep(2)

    # Get all job cards
    jobs = driver.find_elements(By.CLASS_NAME, "jobTitle-link")

    for job in jobs:
        title = job.text.strip()
        link = job.get_attribute('href')
        job_id = link.split("-")[-1].strip("/")

        if any(keyword in title.lower() for keyword in keywords):
            print(f"Title: {title}")
            print(f"Link: {link}")
            print(f"Job ID: {job_id}")
            print("-" * 50)

    # Try to click "Next" page
    try:
        next_btn = driver.find_element(By.CLASS_NAME, "pagination-link--next")
        if "disabled" in next_btn.get_attribute("class"):
            break
        else:
            next_btn.click()
            page_num += 1
            time.sleep(3)
    except:
        print("No more pages or pagination failed.")
        break

driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
import time

# Setup Chrome in headless mode
options = Options()
options.add_argument('--headless')
options.add_argument('--disable-gpu')
options.add_argument('--window-size=1920,1080')

# Initialize WebDriver
driver = webdriver.Chrome(options=options)

# URL to the job search page
base_url = "https://jobs.myflorida.com/search/?q=&sortColumn=referencedate&sortDirection=desc&searchby=location&d=10"

# Keywords to filter job descriptions
keywords = ['data', 'analytics', 'python', 'developer', 'finance', 'sql', 'power bi', 'dashboards']

# List to store matching jobs
matching_jobs = []

driver.get(base_url)
time.sleep(3)

# Iterate through multiple pages
while True:
    job_links = driver.find_elements(By.CSS_SELECTOR, 'a.jobTitle-link')
    job_urls = [link.get_attribute('href') for link in job_links]

    for job_url in job_urls:
        driver.execute_script("window.open('');")
        driver.switch_to.window(driver.window_handles[1])
        driver.get(job_url)
        time.sleep(2)

        page_text = driver.find_element(By.TAG_NAME, 'body').text.lower()
        if any(keyword in page_text for keyword in keywords):
            print(f"✅ Match found: {job_url}")
            matching_jobs.append(job_url)
        else:
            print(f"❌ No match: {job_url}")

        driver.close()
        driver.switch_to.window(driver.window_handles[0])
        time.sleep(1)

    # Check if a "Next" button exists to go to the next page
    try:
        next_page = driver.find_element(By.LINK_TEXT, 'Next')
        next_page.click()
        time.sleep(3)
    except:
        break

driver.quit()

print("\nAll matching jobs:")
for job in matching_jobs:
    print(job)
```
