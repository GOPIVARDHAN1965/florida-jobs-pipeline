```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time

# Keywords you're interested in
keywords = [
    "data", "analyst", "developer", "engineer", "python", "sql",
    "bi", "cloud", "aws", "etl", "software", "finance", "financial",
    "visualization", "power bi", "dashboards", "information specialist", "review specialist", "budget specialist"
]

# Setup Chrome with visible browser (not headless)
options = webdriver.ChromeOptions()
options.add_argument("--start-maximized")

driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

# Open the State of Florida job search page
driver.get("https://jobs.myflorida.com/search/?searchby=location")
time.sleep(3)

page_num = 1
while True:
    print(f"\n--- PAGE {page_num} ---\n")
    time.sleep(2)

    jobs = driver.find_elements(By.CLASS_NAME, "jobTitle-link")

    for job in jobs:
        title = job.text.strip()
        link = job.get_attribute('href')
        job_id = link.split("-")[-1].strip("/")

        # Check if title matches any keyword
        if any(keyword in title.lower() for keyword in keywords):
            print(f"Title: {title}")
            print(f"Link: {link}")
            print(f"Job ID: {job_id}")
            print("-" * 50)

    # Try to go to next page
    try:
        next_btn = driver.find_element(By.LINK_TEXT, str(page_num + 1))
        next_btn.click()
        page_num += 1
        time.sleep(3)
    except:
        print("No more pages or pagination failed.")
        break

driver.quit()
```
